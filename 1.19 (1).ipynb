```python
import pandas as pd

# Load dataset
df = pd.read_csv("swift_currency_tracker_all_reports.csv")

# Check column names
print(df.columns.tolist())

# Convert report month to datetime
df['report_month'] = pd.to_datetime(df['report_month'])

# Find most recent month
latest_month = df['report_month'].max()

print("Most Recent Month:", latest_month)

# Filter data for latest month and Global Payment Share metric
latest_data = df[
    (df['report_month'] == latest_month) &
    (df['metric'].str.contains('payment', case=False, na=False))
]

# Sort by value in descending order
ranking = latest_data.sort_values('value', ascending=False)

# Select relevant columns
ranking_table = ranking[['currency_or_economy', 'value']].copy()

# Rename columns
ranking_table.columns = ['Currency', 'Payment Share (%)']

# Add rank
ranking_table.insert(0, 'Rank', range(1, len(ranking_table) + 1))

# Display top 10
top10 = ranking_table.head(10)

print("\nTop 10 Currencies by Global Payment Share")
print(top10)

# Save to CSV (optional)
top10.to_csv("Top_Currencies_Global_Payment_Share.csv", index=False)
```
