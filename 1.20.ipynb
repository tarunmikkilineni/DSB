```python
import pandas as pd

# Load dataset
df = pd.read_csv("swift_currency_tracker_all_reports.csv")

# Convert report_month to datetime
df['report_month'] = pd.to_datetime(df['report_month'])

# Find latest month
latest_month = df['report_month'].max()

# Filter latest month data
latest_data = df[df['report_month'] == latest_month]

# Find USD global payment share
usd_share = latest_data[
    (latest_data['currency_or_economy'] == 'USD') &
    (latest_data['metric'].str.contains('payment', case=False, na=False))
]

print("Latest Month:", latest_month.strftime("%Y-%m"))

if not usd_share.empty:
    print("USD Global Payment Share:", usd_share['value'].iloc[0], "%")
else:
    print("USD payment share data not found.")
```
