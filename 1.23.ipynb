```python
import pandas as pd

df = pd.read_csv("swift_currency_tracker_all_reports.csv")

# Look at available categories
print(df['category'].unique())

# Filter offshore RMB data
offshore = df[
    df['category'].str.contains('offshore', case=False, na=False)
]

# Sort descending by share
result = offshore.sort_values('value', ascending=False)

# Top economy
top_economy = result.iloc[0]

print(f"Highest Offshore RMB Economy: {top_economy['currency_or_economy']}")
print(f"Share: {top_economy['value']}%")
```
