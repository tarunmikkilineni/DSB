```python
import pandas as pd

# Load dataset
df = pd.read_csv("swift_currency_tracker_all_reports.csv")

# Check available metrics
print(df['metric'].unique())

# Filter payment share records
payment_df = df[df['metric'].str.contains('payment', case=False, na=False)].copy()

# For each month, get Top 5 currencies
top5_each_month = (
    payment_df
    .sort_values(['report_month', 'value'], ascending=[True, False])
    .groupby('report_month')
    .head(5)
)

# Count appearances
currency_counts = (
    top5_each_month['currency_or_economy']
    .value_counts()
    .reset_index()
)

currency_counts.columns = ['Currency', 'Times in Top 5']

# Display result
print("\nCurrencies Appearing Most Often in Top 5")
display(currency_counts)

# Save output (optional)
currency_counts.to_csv("Currency_Top5_Frequency.csv", index=False)
```
