```python
import pandas as pd

# Load dataset
df = pd.read_csv("swift_currency_tracker_all_reports.csv")

# Check available metrics
print(df['metric'].unique())

# Filter for Global Payment Share data
payment_data = df[df['metric'].str.contains('payment', case=False, na=False)]

# Find latest month
latest_month = payment_data['report_month'].max()

# Filter latest month
latest_data = payment_data[payment_data['report_month'] == latest_month]

# Sort by payment share to calculate rank
latest_data = latest_data.sort_values('value', ascending=False).reset_index(drop=True)
latest_data['Rank'] = latest_data.index + 1

# Find CNY/RMB row
cny = latest_data[
    latest_data['currency_or_economy'].str.contains('CNY|RMB|Renminbi|Chinese Yuan',
                                                    case=False,
                                                    na=False)
]

if not cny.empty:
    share = cny['value'].iloc[0]
    rank = cny['Rank'].iloc[0]

    print(f"Latest Month: {latest_month}")
    print(f"CNY/RMB Global Payment Share: {share}%")
    print(f"CNY/RMB Global Rank: {rank}")
else:
    print("CNY/RMB data not found.")
```
