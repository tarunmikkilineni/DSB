```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("swift_currency_tracker_all_reports.csv")

# Find latest month
latest_month = df['report_month'].max()

# Filter payment share data for latest month
latest_data = df[
    (df['report_month'] == latest_month) &
    (df['metric'].str.contains('payment', case=False, na=False))
]

# Get Top 10 currencies
top10 = latest_data.sort_values('value', ascending=False).head(10)

# Create bar chart
plt.figure(figsize=(10,6))
plt.bar(top10['currency_or_economy'], top10['value'])

plt.title(f"Top 10 Currencies by Global Payment Share ({latest_month})")
plt.xlabel("Currency")
plt.ylabel("Payment Share (%)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
